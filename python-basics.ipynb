```python

# Write a program to calculate the area of circle,square & rectangle.
 # Area of circle
pi=3.14
radius = float(input("Enter radius of circle: "))
circle = pi*radius*radius
print("Area of Circle =",circle)

 # Area of square
side = float(input("Enter side of square: "))
square = side * side
print("Area of Square =", square)

 # Area of rectangle
length = float(input("Enter length of rectangle: "))
breadth = float(input("Enter breadth of rectangle: "))
rectangle = length * breadth
print("Area of Rectangle =", rectangle)
```

```python
 # Write a program to check the eligibilityof a candidate to vote in an election such that the voter should be of age 18years or more & should be Indian Citizen are country. If the condition are true then he is eligible.

age = int(input("Enter your age: "))
citizen = input("Are you a citizen of this country? (yes/no): ")

if age >= 18 and citizen == "yes":
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
 # Write a program to calculate factorial number using a for loop.
num=int(input("Enter the number:"))
fact=1
for i in range(1,num+1):
 fact=fact*i
print("factorial is the given number is:", fact)
```
